```python
import pandas as pd
```

```python
# List of capital names (same order used while saving CSVs)
capitals = [
    "Amaravati", "Itanagar", "Dispur", "Patna", "Raipur", "Panaji", "Gandhinagar",
    "Chandigarh", "Shimla", "Ranchi", "Bengaluru", "Thiruvananthapuram", "Bhopal",
    "Mumbai", "Imphal", "Shillong", "Aizawl", "Kohima", "Bhubaneswar", "Jaipur",
    "Gangtok", "Chennai", "Hyderabad", "Agartala", "Lucknow", "Dehradun", "Kolkata"
]



# For each capital city
all_city_data = {}

for city in capitals:
    df_city = None
    for i, param in enumerate(parameters_of_the_model):
        file_path = f"dataset/{city.lower()}_{param}_hourly.csv"
        try:
            df_temp = pd.read_csv(file_path)
        except:
            print(file_path)
            continue;

        # Drop duplicate datetime columns
        if i != 0:
            df_temp.drop(columns='datetime', inplace=True)

        df_city = pd.concat([df_city, df_temp], axis=1) if df_city is not None else df_temp

    all_city_data[city] = df_city
```

```python
all_city_data.keys()
```

```python
# Merge all city dataframes into one for model training
merged_df = pd.concat([df for df in all_city_data.values() if df is not None], ignore_index=True)
print(f"Merged dataframe shape: {merged_df.shape}")
merged_df.head()
```

```python
merged_df.to_csv("dataset/final.csv")
```
