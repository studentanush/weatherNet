API FETCHING

```python
import fetch
```

```python
parameters_of_the_model = [
    "ALLSKY_SFC_SW_DWN",     # Surface shortwave downwelling (total solar)
    "ALLSKY_SFC_SW_DIFF",    # Diffuse shortwave radiation
    "ALLSKY_SFC_SW_DNI",     # Direct normal irradiance
    "SI_TILTED_AVG",         # Tilted surface irradiance (PV specific)
    "TOA_SW_DWN",            # Top-of-atmosphere solar radiation
    "AIRMASS_HR",            # Air mass (optical depth indicator)
    "T2M",                   # Air temperature at 2 meters
    "RH2M",                  # Relative humidity at 2 meters
    "QV2M",                  # Specific humidity at 2 meters
    "PS",                    # Surface pressure
    "WS2M",                  # Wind speed at 2 meters
    "CLOUD_AMT",             # Total cloud amount
    "ALLSKY_SFC_LW_DWN",     # Downward longwave radiation
    "ALLSKY_SRF_ALB",        # Surface albedo (reflectivity)
    "SG_SZA_HR" 
]
```

```python
for i in parameters_of_the_model:
    try:
        fetch.fetch_and_save(i,18.5204,73.8567,"pune")
    except:
        print(f"failed to download {i}")
```
